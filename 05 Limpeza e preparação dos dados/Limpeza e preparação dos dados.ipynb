```python
#filtrando o dataframe orders com status 'delivered', ' shipped'
orders_shipped_delivered = orders[orders['order_status'].isin(['delivered', 'shipped' ])]
# valindando a informacao 
orders_shipped_delivered[ 'order_status'].unique()
```

```python
pd.pivot_table(
    orders_shipped_delivered,
    index="order_status",
    values="order_id",
    aggfunc="count"
)
```

```python
delivered_status_list= []

for index, row in orders_shipped_delivered.iterrows():
    if row["order_delivered_customer_date"]> row["order_estimated_delivery_date"]:
        delivered_status_list.append("atrasado")
    elif row["order_estimated_delivery_date"] >=row["order_delivered_customer_date"]:
        delivered_status_list.append("no tempo")
    else:
        delivered_status_list.append("shipped")

orders_shipped_delivered["delivery_status"]= delivered_status_list

orders_shipped_delivered.head(4)

```
